#En este item trabajaremos con datos de finalización de cursos en una plataforma de educación online.

Cada fila representa a un estudiante que tomó un curso en uno de dos formatos posibles: A o B.

🧩Objetivo: Responder: ¿Qué formato de curso parece más efectivo para que los estudiantes completen el curso?

```python
import pandas as pd
df = pd.read_csv('/content/course_completion.csv')

print(df.head())
print() # salto de línea

print(df.groupby("formato_curso")["completo_curso"].mean())
```

Al analizar solo el promedio global, el formato A parece tener mejor desempeño.
Esta es una conclusión razonable, pero incompleta, ya que no considera posibles segmentaciones.

#Cuando la historia cambia al segmentar
En esta práctica, continuaremos trabajando con el mismo dataset, pero ahora incorporaremos un factor adicional del estudiante para verificar si la conclusión inicial se mantiene.

Objetivo: Comprobar si el formato A es el mejor al analizar por grupos.

Paso 1: Segmentar por nivel de experiencia


```python
import pandas as pd
df = pd.read_csv('/content/course_completion.csv')

print(df.groupby(["nivel_experiencia","formato_curso"])["completo_curso"].mean())
```

Al segmentar los datos, la tendencia se invierte. Aunque el formato A parecía mejor en el promedio global, el formato B funciona mejor en todos los subgrupos, lo que indica una paradoja de Simpson.
